```python
import asyncio
import random
from typing import Optional
from playwright.async_api import async_playwright, Browser, Page, Playwright
from playwright_stealth import Stealth
```

```python
from playwright.sync_api import sync_playwright

def run():
    # Lance le navigateur en arrière-plan (headless=True par défaut)
    # Changez à False pour voir le navigateur s'ouvrir à l'écran
    with sync_playwright() as p:
        browser = p.chromium.launch(headless=False)
        page = browser.new_page()
        
        # Navigation vers l'URL
        print("Navigation en cours...")
        page.goto("https://playwright.dev/python/")
        
        # Récupérer et afficher le titre de la page
        title = page.title()
        print(f"Titre de la page : {title}")
        
        # Prendre une capture d'écran
        page.screenshot(path="playwright_home.png")
        print("Capture d'écran sauvegardée sous 'playwright_home.png'")
        
        # Fermeture propre des ressources
        browser.close()
```

```python
run()
```
